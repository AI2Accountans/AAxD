# 1. Leemos tu DataBook

````python
with open(ruta_archivo, 'r', encoding='utf-8') as f:
    contenido = f.read()
````

# 2. Extraemos MÁGICAMENTE los bloques ocultos de JSON-LD

````python
bloques = contenido.split('```json-ld\n')
json_blocks = [b.split('\n```')[0] for b in bloques[1:]]
````

# 3. Cargamos la Semántica (TerminusDB en memoria)

````python
g = Graph()
for b in json_blocks:
    g.parse(data=b, format='json-ld')
````

# 4. Auditoría SPARQL (Sumar todos los montos de la cuenta)

````python
query = """
    PREFIX gl-cor: <http://www.xbrl.org/int/gl/cor/2015-03-25#>

    SELECT (SUM(?monto) AS ?totalCapital)
    WHERE {
        ?s gl-cor:amount ?monto .
    }
"""
resultados = g.query(query)
print("--- REPORTE DE AUDITORÍA (MOMENTO 0) ---")
for fila in resultados:
    print(f"💰 El Capital Total matemáticamente comprobado es: ${fila.totalCapital} COP")
````
